```python
def prepare_data(df):
    # create a copy of the dataframe
    df = df.copy()
    # isolate the numerical and categoriccal column names we are interested in in a list
    cat= df.select_dtypes(include=["object"]).columns.to_list()
    num = ["engine_size", "mileage"]
    # gets the age of the car | feature engineering
    cutoff_year = df.year.max()
    df["age"] = cutoff_year - df["year"]
    num = num + ["age"]
    # reshape price into a two dimensional array
    y = df["price"].values.reshape(-1,1)
    # drop irrelevant columns
    df_new = df.drop(["year", "price", "car_id"], axis=1)
    # one hot encoding
    
    # encoder = OneHotEncoder(sparse_output=False)
    # one_hot_encoded = encoder.fit_transform(df_new[cat])
    # one_hot_df = pd.DataFrame(one_hot_encoded, columns=encoder.get_feature_names_out(cat))
    # df_new = df_new.drop(cat, axis=1)
    # encoded_df = pd.concat([df_new, one_hot_df], axis=1)
    # # convert the dataframe to an array
    # X = np.hstack((df_new.values, one_hot_df.values))
    X = df_new[num].values
    # return the new dataframe, the X and the target variable 'y'
    return df_new, X, y
```

```python
encoder = OneHotEncoder(sparse_output=False)
one_hot_encoded = encoder.fit_transform(df_new[cat])

one_hot_df = pd.DataFrame(one_hot_encoded, columns=encoder.get_feature_names_out(cat))

df_encoded = pd.concat([df_new, one_hot_df], axis=1)

df_encoded = df_encoded.drop(cat, axis=1)
df_encoded
```
