```python
from crewai import Agent, Task, Crew
```

```python
from crewai_tools import (
  FileReadTool,
  MDXSearchTool,
)

read_resume = FileReadTool(file_path='./fake_resume.md')
semantic_search_resume = MDXSearchTool(mdx='./fake_resume.md')
```

```python
# === Agent 1: CV Creator ===
cv_creator = Agent(
    role="ATS-Compliant CV Creator",
    goal="Create a tailored ATS-friendly CV",
    tools = [read_resume, semantic_search_resume],
    backstory=("You are an expert CV writer with deep ATS compliance knowledge"
            "With a strategic mind and an eye for detail, you "
            "excel at refining resumes to highlight the most "
            "relevant skills and experiences, ensuring they "
            "resonate perfectly with the job's requirements."
        )
)


# === Agent 2: Cover Letter Creator ===
cover_letter_creator = Agent(
    role="Tailored Cover Letter Writer",
    goal="Write a compelling cover letter aligned with CV and job"
        "description",
    tools = [read_resume, semantic_search_resume],
    backstory=(
        "You are a Specialist in writing persuasive,"
        "job-winning cover letters"
        "Equipped with analytical prowess, you dissect "
        "and synthesize information from the job description and resume "
        "to craft a comprehensive "
        "personal and professional cover letter"
    )
)


# === Agent 3: ATS Compliance & Humanization Checker ===
ats_checker = Agent(
    role="ATS Compliance & Human Tone Reviewer",
    goal="Ensure CV & Cover Letter are ATS-compliant,"
        "natural, and human-readable",    
    tools=[read_resume, semantic_search_resume],
    backstory=(
        "Experienced recruiter and ATS analyst ensuring"
        "recruiter-friendly documents"
        "Your skills help pinpoint the necessary "
        "qualifications and skills sought "
        "by employers, forming the foundation for "
        "effective application tailoring."
    ),
    allow_delegation=True
)

# ==== Tasks ===
cv_task = Task(
    description=(
        "Generate an ATS-compliant CV for the given candidate name {candidate_name}."
        "Using the profile and job requirements obtained from the job description {job_description}"
        "tailor the resume to highlight the most "
        "relevant areas. Employ tools to adjust and enhance the "
        "resume content."
        "Update every section, inlcuding the initial"
        "summary, work experience, skills, "
        "and education. All to better enhance the candidates "
        "abilities and how it matches the job description {job_description}"
        "Input: {candidate_name}, {location}, {job_role}, {job_type}, {job_description}. "
        "Output: cv_text in markdown format."
    ),
    expected_output=(
        "A comprehensive and updated resume that effectively highlights the candidate's "
        "qualifications and experiences relevant to the job {job_description}."
    
    ),
    agent=cv_creator
)


cover_letter_task = Task(
    description=(
        "Using the CV or resume from previous task and job description {job_description}"
        "using the tools, create a tailored cover letter. "
        "Ensure tone is professional and matches candidate profile."
    ),
    expected_output=(
        "A comprehensive profile document that includes skills, "
        "project experiences, contributions, interests. to company {company_name}"
    ),
    context=[cv_task],
    agent=cover_letter_creator
)


task_checker = Task(
    description=(
        "Validate both CV and Cover Letter from previous tasks"
        "for ATS compliance, remove AI-like phrases, "
        "ensure human tone, and output as a markdown"
        "Make sure this is the best resume ever but "
        "don't make up any information. 

    ),
    expected_output=(
       "An updated resume that effectively highlights the candidate's "
        "qualifications and experiences relevant to the job"
        "and a cover letter tailored to the job"
        "requirements both formatted as a markdown file."
    ),
    context=[cv_task, cover_letter_task]
    output_file="Resume_and_cover_letter.md",
    agent=ats_checker
)

# === Crew Orchestration ===
crew = Crew(
    agents=[cv_creator, cover_letter_creator, ats_checker],
    tasks=[task_cv, task_cover_letter, task_checker],
    manager_llm=ChatOpenAI(model="gpt-3.5-turbo", 
                           temperature=0.7),
)

expected_inputs = {
    "candidate_name": "John Doe",
    "company_name": "JUNEBusiness",
    "location": "London, UK",
    "job_role": "Data Analyst",
    "job_type": "Hybrid",
    "job_description": '''
As a Data Analyst, you will play a key role in transforming data into insights that drive strategic business decisions. You will work closely with product, marketing, and development teams to gather, analyze, and interpret data that supports the company’s ongoing growth and operational efficiency.
Key Responsibilities

Collect, process, and clean data from various sources to ensure data accuracy and reliability.
Perform deep-dive analysis to identify trends, patterns, and actionable insights.
Develop dashboards and reports that communicate findings to internal stakeholders.
Support product and business teams with ad-hoc analysis and data-driven recommendations.
Collaborate with cross-functional teams to define and monitor key performance indicators (KPIs).
Maintain data integrity and security while optimizing workflows and automation processes.
Contribute to data strategy and governance initiatives to improve overall data maturity.
Requirements

Proven experience as a Data Analyst, Business Analyst, or similar role in a tech-driven environment.
Proficiency in SQL and data visualization tools (e.g., Power BI, Tableau, Looker).
Strong analytical skills with the ability to translate complex data into clear insights.
Experience with Python, R, or other statistical tools is a plus.
Familiarity with cloud-based data platforms (e.g., AWS, Google BigQuery, Snowflake).
Excellent problem-solving abilities and attention to detail.
Strong communication skills, both written and verbal.
Bachelor's degree in Data Science, Statistics, Computer Science, or a related field.''' }
```
